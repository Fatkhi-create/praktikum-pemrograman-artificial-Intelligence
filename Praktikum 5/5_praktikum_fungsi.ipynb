```python
def bilangan(x, y):
    print('Dalam Fungsi')
    print('\nSebelum Proses')
    print('\nNilai X ', x)
    print('\nNilai Y ', y)
    z = x
    x = y
    y = z
    print('\nSetelah Proses')
    print('\nNilai X ', x)
    print('\nNilai Y ', y)
#program utama
x = 12
y = 3
print('Sebelum memanggil Fungsi, X Bernilai ', x)
print('Sebelum memanggil Fungsi, y Bernilai ', y)
bilangan(x, y)
print('Setelah memanggil Fungsi, X Bernilai ', x)
print('Setelah memanggil Fungsi, y Bernilai ', y)
```
