```python
#Contoh penggunaan Nested Loop

i = 2
while(i < 100):
    j = 2
    while(j <= (i/j)):
        if not(i%j): break
        j = j + 1
    if(j > i/j) : print(i, " is prime")
    i = i + 1

print("Good Bye")
```

```python
number = (1, 2, 3, 4, 5, 6, 7, 8, 9)
ganjil = 0 
genap = 0

for x in number:
    if x % 2 == 0:
        genap += 1
    else: 
        ganjil += 1

print("List angka,", number)
print("Jumlah bilangan genap : ",genap)
print("Jumlah bilangan ganjil : ",ganjil)
```
