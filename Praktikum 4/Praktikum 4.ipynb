```python
awal = int(input('Set Nilai Awal : '))
akhir = int(input('Set Nilai Akhir : '))

count = 0
sumn = 0

print('Bilangan antara %d dan %d' %(awal,akhir))

for i in range(awal,akhir+1) :
    print(i, end = ' ')
    count = count + 1
    sumn = sumn + i

print('\nBilangan di atas ada %d bilangan' %count)
print('Jumlah semua bilangan adalah %d ' %sumn)
```

```python
banyakMhs = int(input('Banyak Mahasiswa : '))
for M in range (banyakMhs) :
    print('Mahasiswa ke-', M + 1)
    banyakMK = int(input('Banyak Mata Kuliah yang diambil : '))
    totalnilai = 0
    for N in range (banyakMK) :
        nilai = int(input('Input Nilai ke %d :' %(N + 1)))
        totalnilai = totalnilai + nilai
    rerata = totalnilai / banyakMK
    print('Rata - rata : ',rerata)
```

```python
banyakbil = int(input('Input Banyak bilangan :'))
for i in range(banyakbil + 1):
    bil = int(input('Input bilangan :'))
    if bil > 0 :
        print(bil, ' adalah jenis bilangan positif')
    elif bil == 0:
        print(bil, ' adalah bilangan nol')
    else  :
        print(bil, ' adalah jenis bilangan negatif')
```

```python
n = int(input('Masukkan Banyak Barang : '))

total_pembelian = 0

for i in range(1, n + 1):
    print("Barang ke-", i)
    harga = int(input("Harga Barang : "))
    jumlah = int(input("Jumlah Barang : "))

    total_pembelian = total_pembelian + (harga * jumlah)
    print()

print("Total Pembelian adalah Rp", total_pembelian)
```

```python
n = int(input("Input Banyak Bilangan :"))

genap = 0
ganjil = 0

for i in range (1, n + 1):
    bill = int(input("Input Bilangan :"))

    if bill % 2 == 0 :
        genap = genap + 1
    else :
        ganjil = ganjil + 1

print("Banyak Bilangan Genap : ",genap)
print("Banyak Bilangan Ganjil : ",ganjil)
```
